```python
import pandas as pd
```

```python
# load spotify_reviews.csv
df = pd.read_csv('spotify_reviews.csv')
```

```python
columns_to_drop = ['reviewId', 'userName', 'score', 'thumbsUpCount', 'reviewCreatedVersion']
df = df.drop(columns=columns_to_drop)
```

```python
# filter for all reviews after 2023-01-01
df['at'] = pd.to_datetime(df['at'])
df = df[df['at'] > '2023-01-01']

# confirm date range
print(df['at'].min())
print(df['at'].max())

# save to json
df.sample(1000)
df.to_json('spotify_reviews_post-2023_1000.json', orient='records')
```
