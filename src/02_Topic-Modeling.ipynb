# BERTopic Modeling

### Loading necessary libraries

```python
import pandas as pd
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from bertopic.representation import MaximalMarginalRelevance
```

## EDA & Preprocessing

### Loading and Previewing Dataset

```python
# Define the path to the CSV file
file_path = "../data/spotify_reviews_clean&PP_NoEmoji_V2.csv"

# Load the dataset from the specified CSV file using pandas
data = pd.read_csv(file_path)

# Display the first five rows of the dataframe to check the data's initial structure
data.head()
```

### Data Preprocessing

```python
# Extracting reviews into a list from the 'content' column
reviews = data['content'].tolist()
```

## Initializing Topic Modeling

```python
# Creating a topic model with detailed settings

# Load the all-MiniLM-L6-v2 model from sentence-transformers
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
# Load the MaximalMarginalelevance model to increase granularity
representation_model = MaximalMarginalRelevance(diversity=0.7)

# Initialize BERTopic
topic_model = BERTopic(language="english", calculate_probabilities=True,
                       verbose=True, representation_model=representation_model,
                       top_n_words=10, embedding_model=embedding_model, min_topic_size=10)
```

## Topic Modeling on all Reviews

### Fitting the Topic Model to Reviews

```python
# Generating topics and their probabilities
topics, probabilities = topic_model.fit_transform(reviews)
```

```python
# Getting information about the topics found in the dataset
topic_info = topic_model.get_topic_info()
```

### Saving the Fitted Model

```python
topic_model.save("spotify_reviews_topic_model")
```

### Topics Over Time

```python
# Ensure timestamps are datetime objects
data['at'] = pd.to_datetime(data['at'])
timestamps = data['at'].tolist()

# Set the number of bins for temporal analysis
nr_bins = 10  # Adjust this based date range and desired granularity

# Generate the topics over time data using the parameters
topics_over_time = topic_model.topics_over_time(docs=reviews, topics=topics, timestamps=timestamps, nr_bins=nr_bins)

# Visualize the topics over time
topic_model.visualize_topics_over_time(topics_over_time, top_n_topics= 10)
```
